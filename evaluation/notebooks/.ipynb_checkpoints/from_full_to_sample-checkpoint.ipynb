```python
import pandas as pd
import os
import json
import tqdm

# Paths
input_folder = '../outputs-original-full'
output_folder = '../outputs'
os.makedirs(output_folder, exist_ok=True)  # make sure output folder exists

# Load your sampled test CSV
df = pd.read_csv('../../dataset/test_data.csv')
df_sample = df.sample(n=100, random_state=42)
print(df_sample)
# Iterate through all JSONL files in the original folder
for filename in os.listdir(input_folder):
    if filename.endswith('.jsonl'):
        input_path = os.path.join(input_folder, filename)
        output_path = os.path.join(output_folder, filename)
        filtered_results = pd.DataFrame()
        df_results = pd.read_json(input_path, lines=True)
        for index in tqdm.tqdm(range(len(df_sample))):
            row = df_sample.iloc[index]
            # Use your correct access method
            print(df_results["Row Number"])
            curr_row = df_results[df_results["Row Number"] == int(row["Row Number"])].iloc[0]
            filtered_results = pd.concat([filtered_results, pd.DataFrame([curr_row])], ignore_index=True)

        filtered_results.to_json(output_path, orient='records', lines=True)


        print(f"Saved filtered results for {filename} with original formatting and sample order")
```

```python
import os
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

results_dir = "../results-original-full"  # adjust if needed

# Collect data
data_list = []

for fname in os.listdir(results_dir):
    if fname.endswith(".json"):
        fpath = os.path.join(results_dir, fname)
        with open(fpath, "r") as f:
            data = json.load(f)
            avg = data.get("overall", {}).get("average", None)
            if avg is not None:
                # Assuming filename format like: results_MODEL_CONDITION_SIZE.json
                parts = fname.replace(".json", "").split("_")
                
                # Extract fields
                model_name = parts[1]        # e.g., gpt-oss-20b
                condition = "original"
                if "empty" in fname:
                    condition = "empty"
                if "gpt-oss-20b" in model_name.lower():
                    if "low" in fname:
                        condition += "_low"
                    if "medium" in fname:
                        condition += "_medium"
                    if "high" in fname:
                        condition += "_high"
                size = "sample"
                if "full" in fname:
                    size = "full"

                data_list.append({
                    "Model": model_name,
                    "Condition": condition,
                    "Size": size,
                    "Average": avg
                })

# Convert to DataFrame
df = pd.DataFrame(data_list)

df["Model_short"] = df["Model"].apply(get_short_name)

# Create simple condition for coloring
df["Condition_simple"] = df["Condition"].apply(
    lambda x: "original" if x.startswith("original") else "empty"
)

def get_short_name(model_full):
    if "dapo-qwen" in model_full.lower():
        return "DAPO-Q32B"
    elif "qwq-32b" in model_full.lower():
        return "QwQ-32B"
    elif "gpt-oss-20b" in model_full.lower():
        return "OSS-20B"
    elif "nemotron" in model_full.lower():
        return "NRR-Q1.5B"
    elif "openthinker" in model_full.lower():
        return "OpenT-7B"
    else:
        return model_full  # fallback: keep original if unknown


df["Model_short"] = df["Model"].apply(get_short_name)

# Custom model order
model_order = ["NRR-Q1.5B", "OpenT-7B", "OSS20B", "QwQ-32B", "DAPO-Q32B"]

# Color palette: blue for empty, pastel greens for original variations
unique_conditions = df['Condition'].unique()
empty_conditions = [c for c in unique_conditions if not c.startswith("original")]
original_conditions = [c for c in unique_conditions if c.startswith("original")]

empty_color = "#89CFF0"  # pastel blue
greens = sns.color_palette("Greens", n_colors=len(original_conditions), desat=0.6)

color_map = {c: empty_color for c in empty_conditions}
for c, g in zip(original_conditions, greens):
    color_map[c] = g

# Sort hue order: empty first, then original variations
hue_order = empty_conditions + original_conditions

# Seaborn settings
sns.set(style="whitegrid", font_scale=1.1)

# Horizontal bar plot faceted by Size
g = sns.catplot(
    data=df,
    y="Model_short",
    x="Average",
    hue="Condition",
    col="Size",
    kind="bar",
    height=6,
    aspect=1.2,
    palette=color_map,
    sharex=True,
    order=model_order,
    hue_order=hue_order,
    dodge=True
)

# Add numbers on bars and visual cue for OSS-20B multiple modes
for ax in g.axes.flatten():
    for p in ax.patches:
        width = p.get_width()
        if width > 0:
            ax.text(width + 0.5,
                    p.get_y() + p.get_height() / 2,
                    f"{width:.2f}",
                    ha='left', va='center', fontsize=9)
    # Visual cue for OSS-20B multiple original modes
    if "OSS20B" in model_order:
        ax.annotate("3 original modes",
                    xy=(0, model_order.index("OSS20B")),
                    xytext=(-0.5, model_order.index("OSS20B")),
                    va='center', ha='right', fontsize=9, color='gray')

# Legend on top-right outside plot
g._legend.set_title("Condition")
g._legend.set_bbox_to_anchor((1.05, 0.95))
g._legend.set_frame_on(False)

# Axis labels and titles
g.set_titles("Size: {col_name}")
g.set_axis_labels("Overall Average", "Model")

plt.tight_layout()
plt.show()
```
