```python
import os
import json
import re
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
import warnings
warnings.filterwarnings('ignore')

plt.rcParams.update({'font.size': 10})

MODEL_NAMES = {
    'openai_gpt-oss-20b': 'oss',
    'BytedTsinghua-SIA_DAPO-Qwen-32B': 'dapo', 
    'Qwen_QwQ-32B': 'qwq',
    'open-thoughts_OpenThinker-7B': 'opent',
    'nvidia_Nemotron-Research-Reasoning-Qwen-1.5B': 'nrr'
}

COLORS = {
    'empty': '#E8E8E8',
    'default': '#8FBC8F', 
    'sampled': '#F0B7A4',
    'transfer': '#A8DADC',
    'ensemble': '#B8A9C9'
}

def parse_jsonl(filepath):
    """Parse JSONL file."""
    try:
        with open(filepath, 'r') as f:
            return [json.loads(line) for line in f if line.strip()]
    except:
        return []

def get_variation(filename):
    """Extract thought variation from filename."""
    # Check for without_answer flag first
    without_ans = 'without_answer' in filename
    
    if 'zero_shot_empty.jsonl' in filename:
        return 'empty'
    
    if ('_zero_shot_original.jsonl' in filename or '_zero_shot_original_low.jsonl' in filename):
        return 'default without ans' if without_ans else 'default full text'
    
    if '_with_sampling.jsonl' in filename:
        return 'sampled without ans' if without_ans else 'sampled full text'
    
    # Transfer CoT
    match = re.match(r'(.+)_thoughts_to_(.+)_zero_shot', filename)
    if match:
        source = MODEL_NAMES.get(match.group(1), match.group(1)).lower()
        suffix = 'without ans' if without_ans else 'full text'
        return f'transfer cot {source} {suffix}'
    
    # Ensemble CoT
    if '_gen_' in filename and '_eval_' in filename:
        gen = re.search(r'_gen_([^_]+)', filename)
        eval_match = re.search(r'_eval_([^_]+)', filename)
        if gen and eval_match:
            gen_short = MODEL_NAMES.get(gen.group(1), gen.group(1)).lower()
            eval_short = MODEL_NAMES.get(eval_match.group(1), eval_match.group(1)).lower()
            suffix = 'without ans' if without_ans else 'full text'
            return f'ensemble cot {gen_short}/{eval_short} {suffix}'
    
    return None

def check_consistency(dp1, dp2, is_thoughts_to):
    """Check consistency (agreement) between two data points."""
    fields = ("Target Answer", "Target Result") if is_thoughts_to else ("LLM Answer", "Result")
    
    ans1, res1 = dp1.get(fields[0], ""), dp1.get(fields[1], "")
    ans2, res2 = dp2.get(fields[0], ""), dp2.get(fields[1], "")
    
    invalid = ["not defined", "N/A", "does not match", "are not permitted"]
    if any(p in ans1 or p in ans2 for p in invalid):
        return None
    
    # Models are consistent if:
    # 1. They give exactly the same answer (regardless of correctness)
    # 2. OR both are correct (even if answers differ slightly but both marked correct)
    if ans1 == ans2:
        return 1
    elif res1 == "Correct" and res2 == "Correct":
        return 1
    else:
        return 0

def get_target_model(filename):
    """Extract target model from filename."""
    # For transfer CoT files, extract the target (after thoughts_to_)
    if '_thoughts_to_' in filename:
        match = re.match(r'.+_thoughts_to_(.+)_zero_shot', filename)
        if match:
            return match.group(1)
    
    # For other files, find any model name in the filename
    for model in MODEL_NAMES:
        if model in filename:
            return model
    return None

def calculate_pairwise_consistency(folder_path):
    """Calculate pairwise consistency for each variation."""
    if not os.path.exists(folder_path):
        return {}
    
    variation_files = defaultdict(list)
    
    print("Processing files:")
    for filename in os.listdir(folder_path):
        if not filename.endswith('.jsonl'):
            continue
            
        variation = get_variation(filename)
        target = get_target_model(filename)
        
        print(f"  {filename} -> variation: {variation}, target: {target}")
        
        if variation and target:
            variation_files[variation].append((target, filename))
    
    print(f"\nGrouped variations:")
    for var, files in variation_files.items():
        print(f"  {var}: {len(files)} files")
    
    results = {}
    
    for variation, files in variation_files.items():
        # if len(files) < 2:
        #     print(f"  Skipping {variation} - only {len(files)} file(s)")
        #     continue
            
        rates = []
        is_thoughts_to = '_thoughts_to_' in files[0][1]
        
        for i in range(len(files)):
            for j in range(i + 1, len(files)):
                data1 = parse_jsonl(os.path.join(folder_path, files[i][1]))
                data2 = parse_jsonl(os.path.join(folder_path, files[j][1]))
                
                if not data1 or not data2:
                    continue
                
                consistent = total = 0
                for dp1, dp2 in zip(data1, data2):
                    result = check_consistency(dp1, dp2, is_thoughts_to)
                    if result is not None:
                        total += 1
                        consistent += result
                
                if total > 0:
                    rates.append(consistent / total)
                    print(f"    {variation} {files[i][0]} vs {files[j][0]}: {consistent}/{total} = {consistent/total:.3f}")
        
        if rates:
            results[variation] = {
                'mean': np.mean(rates),
                'std': np.std(rates),
                'count': len(rates)
            }
    
    return results

def plot_results(results):
    """Plot pairwise consistency results."""
    if not results:
        return []
    
    sorted_items = sorted(results.items(), key=lambda x: x[1]['mean'], reverse=True)
    
    fig, ax = plt.subplots(figsize=(15, 8))
    
    names = [item[0] for item in sorted_items]
    means = [item[1]['mean'] for item in sorted_items]
    stds = [item[1]['std'] for item in sorted_items]
    counts = [item[1]['count'] for item in sorted_items]
    
    colors = []
    for name in names:
        if 'empty' in name:
            colors.append(COLORS['empty'])
        elif 'default' in name:
            colors.append(COLORS['default'])
        elif 'sampled' in name:
            colors.append(COLORS['sampled'])
        elif 'transfer' in name:
            colors.append(COLORS['transfer'])
        elif 'ensemble' in name:
            colors.append(COLORS['ensemble'])
        else:
            colors.append('#E2E2DF')
    
    bars = ax.bar(range(len(names)), means, yerr=stds, color=colors, alpha=0.8, 
                  capsize=3, edgecolor='white', linewidth=1)
    
    ax.set_xlabel('Thought Variation', fontweight='bold')
    ax.set_ylabel('Pairwise Consistency Rate', fontweight='bold')
    ax.set_title('Pairwise Target Consistency Across Thought Variations', fontweight='bold')
    ax.set_xticks(range(len(names)))
    ax.set_xticklabels(names, rotation=45, ha='right', fontsize=9)
    ax.grid(axis='y', alpha=0.3)
    ax.set_ylim(0, 1.0)
    
    for bar, mean, count in zip(bars, means, counts):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
                f'{mean:.3f}\n(n={count})', ha='center', va='bottom', 
                fontsize=8, fontweight='bold')
    
    from matplotlib.patches import Patch
    legend = [Patch(color=color, label=label.title()) for label, color in COLORS.items()]
    ax.legend(handles=legend, loc='upper right')
    
    plt.tight_layout()
    plt.savefig('pairwise_consistency.pdf', bbox_inches='tight')
    plt.show()
    
    return sorted_items

def print_summary(sorted_items):
    """Print analysis summary."""
    print("\nPAIRWISE TARGET CONSISTENCY ANALYSIS")
    print("=" * 60)
    
    for i, (name, data) in enumerate(sorted_items):
        print(f"{i+1:2d}. {name:40s} {data['mean']:.3f} ± {data['std']:.3f} (n={data['count']})")
    
    if sorted_items:
        best = sorted_items[0]
        worst = sorted_items[-1]
        print(f"\nBest: {best[0]} ({best[1]['mean']:.3f})")
        print(f"Worst: {worst[0]} ({worst[1]['mean']:.3f})")
        print(f"Gap: {best[1]['mean'] - worst[1]['mean']:.3f}")

def main(folder_path):
    """Run pairwise consistency analysis."""
    print("Pairwise Target Consistency Analysis")
    print(f"Folder: {folder_path}")
    print("-" * 40)
    
    results = calculate_pairwise_consistency(folder_path)
    
    if not results:
        print("No valid comparisons found.")
        return
    
    print(f"Found {len(results)} variations")
    
    sorted_items = plot_results(results)
    print_summary(sorted_items)
    print("\nAnalysis complete.")

if __name__ == "__main__":
    main("../outputs")
```
